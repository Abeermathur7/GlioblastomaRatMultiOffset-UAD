```python
import numpy as np
```

```python
zspec_metrics = np.load('data/FitMetrics_ZSpec_OptimThresh.npz', allow_pickle=True)
params_metrics = np.load('data/FitMetrics_Params_OptimThresh.npz', allow_pickle=True)

print(zspec_metrics.files)
print(params_metrics.files)
```

```python
zspec_metrics['IF'].item().keys()
```

```python
metric_dict = []
for file in zspec_metrics.files:
    model_name = file
    model_metrics = zspec_metrics[file].item()
    
    model_dict = {
        'Model name': model_name,
        'Accuracy': model_metrics['Accuracy'],
        'F1 score': model_metrics['F1_score'],
        'ROC-AUC': model_metrics['ROC-AUC'],
        'PR-AUC': model_metrics['PR-AUC'],
        'Precision': model_metrics['Precision'],
        'Recall': model_metrics['Recall']
    }
    
    metric_dict.append(model_dict)
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(metric_dict)
df
```
